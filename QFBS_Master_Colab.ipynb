```python
from qfbs.models.poisson import score_matrix, market_probabilities
from qfbs.betting.value import expected_value, kelly_fraction

home_lambda = 1.55  # placeholder; estimate from historical data before use
away_lambda = 1.10  # placeholder; estimate from historical data before use
matrix = score_matrix(home_lambda, away_lambda, max_goals=12)
markets = market_probabilities(matrix)
markets
```

```python
# Example only: replace with a real, timestamped bookmaker price.
market_probability = markets['home']
decimal_odds = 2.20
print('EV per 1-unit stake:', round(expected_value(market_probability, decimal_odds), 4))
print('Quarter-Kelly fraction:', round(kelly_fraction(market_probability, decimal_odds, fraction=0.25), 4))
```
